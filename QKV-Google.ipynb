```python
import os
from google.colab import userdata

os.environ['KAGGLE_USERNAME'] = userdata.get('KAGGLE_USERNAME')
os.environ['KAGGLE_KEY'] = userdata.get('KAGGLE_KEY')
os.environ['KERAS_BACKEND'] = 'jax'

import keras, keras_hub

tokenizer = keras_hub.models.Gemma3Tokenizer.from_preset('gemma3_1b')
model = keras_hub.models.Gemma3Backbone.from_preset('gemma3_1b', dtype='float32')
```

```python
sentence = 'The dog did not cross the road because it was'
```

```python
import numpy as np

ids = [tokenizer.start_token_id] + [int(t) for t in keras.ops.convert_to_numpy(tokenizer(sentence))]
inputs = {'token_ids': np.array([ids]), 'padding_mask': np.ones((1, len(ids)), dtype='int32')}

tokens = [tokenizer.id_to_token(tid).replace('\u2581', ' ') for tid in inputs['token_ids'][0]]

it_idx = tokens.index(' it')
dog_idx = tokens.index(' dog')
road_idx = tokens.index(' road')
```

## Tokens become vectors​

```python
from types import SimpleNamespace

outputs = SimpleNamespace(last_hidden_state=keras.ops.convert_to_numpy(model(inputs)))

for word, idx in [('dog', dog_idx), ('road', road_idx), ('it', it_idx)]:
    vec = outputs.last_hidden_state[0, idx][:5]
    values = ', '.join(f'{v:+.3f}' for v in vec.tolist())
    print(f'{word} = [{values}, ...]')
```

## From output vector to next token prediction​

```python
import numpy as np

h = keras.ops.convert_to_numpy(model(inputs))[0, -1]

print(f'Output vector: [{", ".join(f"{x:+.3f}" for x in h[:8].tolist())}, ...]   ({len(h)} numbers)\n')

top_n_tokens = 100

logits = h @ keras.ops.convert_to_numpy(model.token_embedding.embeddings).T
vocab_size = model.vocabulary_size
probs  = np.exp(logits - logits.max()); probs = probs / probs.sum()
top    = np.argsort(probs)[::-1][:top_n_tokens]

print(f'Scored against all {vocab_size} tokens. top {top_n_tokens}:\n')
for i in top:
    print(f'  {tokenizer.detokenize([int(i)]):<14} {probs[i]*100:5.5f}%')
```

## QK: does the ask match the offer?​

```python
import numpy as np
from keras import ops
from types import SimpleNamespace

qkv_store = {}
attentions = []
hidden_states = []

blocks = model.transformer_layers
n_q, n_kv = model.num_query_heads, model.num_key_value_heads

# Keras has no forward hooks, so we run the layers one by one
# and read Q, K, V out of each attention layer on the way.
x = model.token_embedding(inputs['token_ids']) * ops.sqrt(ops.cast(model.hidden_dim, 'float32'))

for L, block in enumerate(blocks):
    hidden_states.append(ops.convert_to_numpy(x))

    attn = block.attention
    xn = block.pre_attention_norm(x)
    Q = attn._apply_rope(attn.query_norm(attn.query_dense(xn)), 0)
    K = attn._apply_rope(attn.key_norm(attn.key_dense(xn)), 0)
    V = attn.value_dense(xn)
    qkv_store[L] = tuple(ops.convert_to_numpy(t)[0] for t in (Q, K, V))   # (tokens, heads, head_dim)

    Qn, Kn, _ = qkv_store[L]
    T = Qn.shape[0]
    scores = np.einsum('thd,shd->hts', Qn, np.repeat(Kn, n_q // n_kv, axis=1)) / np.sqrt(model.head_dim)
    scores = np.where(np.tril(np.ones((T, T), bool)), scores, -np.inf)
    w = np.exp(scores - scores.max(-1, keepdims=True))
    attentions.append((w / w.sum(-1, keepdims=True))[None])             # (1, heads, tokens, tokens)

    x = block(x, padding_mask=inputs['padding_mask'])

hidden_states.append(ops.convert_to_numpy(x))
outputs = SimpleNamespace(attentions=attentions)
```

```python
import numpy as np

L, H   = 4, 3
d_head = model.head_dim
Q, K, _ = qkv_store[L]
kv     = H // (n_q // n_kv)

q_it   = Q[it_idx,   H]
k_dog  = K[dog_idx,  kv]
k_road = K[road_idx, kv]

N = 5

def row(name, v):
    print(f'  {name:<9} = [{", ".join(f"{x:+.3f}" for x in v[:N].tolist())}, ... ]   ({len(v)} numbers)')

print(f'Gemma 3\n')
row('q_it',   q_it)
row('k_dog',  k_dog)
row('k_road', k_road)

for nm, k in [('dog', k_dog), ('road', k_road)]:
    print(f'\n  q_it · k_{nm}')
    print('    ' + ' + '.join(f'({q_it[i]:+.3f})({k[i]:+.3f})' for i in range(N)) + ' + ...')
    dot = np.dot(q_it, k).item()
    print(f'    = {dot:+.4f}      (all {len(q_it)} terms)')
```

## The full trace​

```python
print('Gemma 3 loaded.')
print(f'Layers : {model.num_layers}')
print(f'Heads  : {model.num_query_heads}  (query heads, sharing {model.num_key_value_heads} key/value head(s))')
print(f'Embedding dimension: {model.hidden_dim}')
print(f'Q, K, V per head: {model.head_dim}')
```

```python
def fmt(v, n=4):
    return '[' + ', '.join(f'{x:+.3f}' for x in v[:n].tolist()) + ', ...]'

def fmt_ends(v, n=3):
    h = ', '.join(f'{x:+.3f}' for x in v[:n].tolist())
    t = ', '.join(f'{x:+.3f}' for x in v[-n:].tolist())
    return f'[{h}, ... , {t}]   (len {len(v)})'

hs = hidden_states

n_layers = model.num_layers
n_heads  = model.num_query_heads
d_head   = model.head_dim
prior    = list(range(it_idx + 1))
names    = [tokens[i].strip() for i in prior]

print('='*100)
print('Starting vector for "it"  (vector embedding x sqrt(d_model); position is added inside attention by RoPE)')
print('='*100)
print(f'  x_it = {fmt(hs[0][0, it_idx], 10)}   ({model.hidden_dim} numbers total)')
print('\n  note: each layer also has an MLP sublayer that adds to x_it.')
print('        the tables below trace the attention half only.\n')

for L in range(n_layers):
    print('\n' + '='*100)
    print(f'LAYER {L}')
    print('='*100)

    hdr = f'{"Head":<6}' + ''.join(f'{n:>10}' for n in names)
    print(hdr); print('-'*len(hdr))
    for H in range(n_heads):
        w = outputs.attentions[L][0][H][it_idx][:len(prior)]
        print(f'{H:<6}' + ''.join(f'{v:>10.3f}' for v in w.tolist()))

    _, _, V = qkv_store[L]
    head_outs = []
    for H in range(n_heads):
        w  = outputs.attentions[L][0][H][it_idx]
        Vh = V[:, H // (n_q // n_kv)]
        head_outs.append((w[:, None] * Vh).sum(0))

    # ---- show the formula with real numbers for head 0 ----
    w0  = outputs.attentions[L][0][0][it_idx]
    V0  = V[:, 0]
    print(f'\n  head output = sum of (attention weight x value vector)')
    print(f'  head 0, written out:\n')
    for i in prior:
        print(f'      {w0[i]:+.3f}  x  v_{names[i]:<8} {fmt(V0[i], 3)}')
    print(f'      {"-"*54}')
    print(f'      sum     = {fmt(head_outs[0], 3)}   ({d_head} numbers)')

    z = np.concatenate(head_outs)
    print(f'\n  all {n_heads} heads flattened:')
    print(f'      z = {fmt_ends(z)}')

    Wo = keras.ops.convert_to_numpy(blocks[L].attention.output_dense.kernel).reshape(n_heads * d_head, -1)

    print(f'\n  W^O  — one per layer, shared by all {n_heads} heads   (top-left 4x4 of {Wo.shape[0]}x{Wo.shape[1]})')
    for r in range(4):
        print('      ' + ''.join(f'{Wo[r, c]:>9.3f}' for c in range(4)))

    attn_out = z @ Wo
    print(f'\n  z @ W^O  — each output slot is z dotted with one column of W^O')
    terms = ' + '.join(f'({z[k]:+.3f})({Wo[k,0]:+.3f})' for k in range(3))
    print(f'      slot 0 = {terms} + ...')
    print(f'             = {attn_out[0]:+.3f}')
    print(f'      attention output = {fmt(attn_out)}')

    attn_out = keras.ops.convert_to_numpy(blocks[L].post_attention_norm(attn_out[None, None]))[0, 0]
    print(f'      after post-attention norm = {fmt(attn_out)}')

    x_old = hs[L][0, it_idx]
    print(f'\n  new x_it = old x_it + attention output')
    print(f'      old x_it     {fmt(x_old)}')
    print(f'    + attn output  {fmt(attn_out)}')
    print(f'    ' + '-'*46)
    print(f'    = new x_it     {fmt(x_old + attn_out)}')
```

```python
import numpy as np, matplotlib.pyplot as plt

n_layers = model.num_layers
n_heads  = model.num_query_heads
prior    = list(range(it_idx + 1))
names    = [tokens[i].strip() for i in prior]

fig, axes = plt.subplots((n_layers + 1) // 2, 2, figsize=(16, 2.5 * n_layers))
axes = axes.flatten()

for L in range(n_layers):
    ax   = axes[L]
    grid = np.zeros((n_heads, len(prior)))
    for H in range(n_heads):
        grid[H] = outputs.attentions[L][0][H][it_idx][:len(prior)]

    im = ax.imshow(grid, aspect='auto', cmap='YlOrRd', vmin=0, vmax=1)
    ax.set_xticks(range(len(prior)))
    ax.set_xticklabels(names, rotation=30, ha='right', fontsize=10)
    ax.set_yticks(range(n_heads))
    ax.set_yticklabels(range(n_heads), fontsize=9)
    ax.set_title(f'Layer {L}', fontsize=12, pad=8)
    if L % 2 == 0:
        ax.set_ylabel('Head', fontsize=10)

    for col, c in [(dog_idx, 'green'), (road_idx, 'red')]:
        ax.axvline(col, color=c, lw=1.3, ls='--', alpha=0.55)

fig.suptitle(f'Attention FROM "it" TO every previous token — all {n_layers} layers x {n_heads} heads',
             fontsize=15, y=0.999)
fig.colorbar(im, ax=axes.tolist(), label='attention weight', shrink=0.4)
plt.savefig('all_layers_heatmap.png', dpi=140, bbox_inches='tight')
plt.show()
```
